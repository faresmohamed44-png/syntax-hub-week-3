```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
import urllib.request

# 1. تحميل نموذج كشف الوجوه الجاهز من OpenCV
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# 2. تحميل صورة تجريبية مستقرة تحتوي على وجوه متعددة
url = "https://raw.githubusercontent.com/opencv/opencv/4.x/samples/data/lena.jpg"
req = urllib.request.urlopen(url)
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
frame = cv2.imdecode(arr, -1)

# تحويل الصورة إلى الأبيض والأسود لتسريع المعالجة
gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

# 3. كشف الوجوه المتعددة
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

# 4. رسم مربعات وتسميات لكل الوجوه المكتشفة
for i, (x, y, w, h) in enumerate(faces):
    cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)
    label = f"Face {i+1}"
    cv2.putText(frame, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

print(f"[INFO] Found {len(faces)} faces!")

# 5. عرض النتيجة النهائية
cv2_imshow(frame)
```
